### Finding HCF of two numbers

```python
def hcf(x,y):
    if x>y:
        s=y
    else:
        s=x
    for i in range(1,s+1):
        if (x%i==0 and y%i==0):
            hcf=i
    print('HCF is: ',hcf)
```

```python
hcf(3,15)
```

```python
def hcf(x,y,z):
    if x>y and z>y:
        s=y
    elif y>x and z>x:
        s=x
    else:
        s=z
    for i in range(1,s+1):
        if (x%i==0 and y%i==0 and z%i==0):
            hcf=i
    print('HCF is: ',hcf)
```

```python
hcf(4,7,15)
```

### Finding LCM of two numbers

```python
def lcm(x,y):
    if x>y:
        g=x
    else:
        g=y
    while True:
        if g%x==0 and g%y==0:
            lcm=g
            break
        g=g+1
    print('LCM is: ',lcm)
    
            
```

```python
lcm(12,9)
```

```python
def lcm(x,y,z):
    if x>y and x>z:
        g=x
    elif y>x and y>z:
        g=y
    else:
        g=z
    while True:
        if g%x==0 and g%y==0 and g%z==0:
            lcm=g
            break
        g=g+1
    print('LCM is: ',lcm)
    
```

```python
lcm(2,4,8)
```

```python
import re

def is_strong_password(password):
    # At least 8 characters long
    if len(password) < 8:
        print('Password length must be 8 characters')
    
    # Contains at least one uppercase letter
    if not re.search("[A-Z]", password):
        print('Password must contain atleast one upper case letter')
    
    # Contains at least one lowercase letter
    if not re.search("[a-z]", password):
        print('Password must contain atleast one lower case letter')
    
    # Contains at least one digit
    if not re.search("[0-9]", password):
        print('Password must contain atleast one digit')
    
    # Contains at least one special character
    if not re.search("[!@#$%^&*()-_+=]", password):
        return False
    return True

def is_weak_password(password):
    # Check if password is the same as "password" (case-insensitive)
    if password.lower() == "password":
        return True
    
    # Check if password is the same as "123456" or "qwerty" (case-insensitive)
    if password.lower() == "123456" or password.lower() == "qwerty":
        return True


# Example usage:
password1 = "Strong!P4ssw0rd"
password2 = "weakpassword"
password3 = "123456"

if is_strong_password(password1):
    print("Password 1 is strong.")
else:
    print("Password 1 is weak.")

if is_weak_password(password2):
    print("Password 2 is weak.")
else:
    print("Password 2 is not weak.")

if is_weak_password(password3):
    print("Password 3 is weak.")
else:
    print("Password 3 is not weak.")
```
